## 03. Convert amplicon frequency files to krona format

```python
def convert_amp_tax_freqs_to_krona(tax_freq, marker_gene):
    krona = tax_freq.drop(['%s sample' % marker_gene], axis=1)
    krona_derep = krona.groupby(['Domain', 'Phylum', 'Class', 'Order', 'Family', 'Genus', 'Species']).sum().reset_index()
    krona_derep['fraction'] = krona_derep['%s reads' % marker_gene] / 218
    krona_derep = krona_derep.drop('%s reads' % marker_gene, axis=1)
    krona_df = krona_derep.rename(columns = {'Domain' : 'superkingdom',
                                             'Phylum' : 'phylum',
                                             'Class' : 'class',
                                             'Order' : 'order',
                                             'Family' : 'family',
                                             'Genus' : 'genus',
                                             'Species' : 'species'})
    krona_df = krona_df[['fraction', 'superkingdom', 'phylum', 'class', 'order', 'family', 'genus', 'species']]
    return(krona_df)
```

### Correct taxonomy of haptophytes, chlorophytes, and oomycota

```python
def correct_taxonomy(df):
    df.loc[:, 'class'] = np.where(df['o'] == 'Prymnesiales', 'Prymnesiophyceae', df['class'])
    df.loc[:, 'class'] = np.where(df['order'] == 'Isochrysidales', 'Prymnesiophyceae', df['class'])
    df.loc[:, 'class'] = np.where(df['order'] == 'Phaeocystales', 'Prymnesiophyceae', df['class'])
    df.loc[:, 'class'] = np.where(df['order'] == 'Peronosporales', 'Peronosporales', df['class'])
    df.loc[:, 'class'] = np.where(df['order'] == 'Saprolegniales', 'Saprolegniales', df['class'])
    df.loc[:, 'class'] = np.where(df['order'] == 'Lagenidiales', 'Lagenidiales', df['class'])
    df.loc[:, 'class'] = np.where(df['order'] == 'Pythiales', 'Pythiales', df['class'])
    df.loc[:, 'order'] = np.where(df['family'] == 'Pycnococcaceae', 'Pseudoscourfieldiales', df['family'])
    df.loc[:, 'class'] = np.where(df['order'] == 'Pseudoscourfieldiales', 'Pseudoscourfieldiophyceae', df['class'])
    return(df)
```
